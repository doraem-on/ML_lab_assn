```python
import numpy as np
x = np.array([1,2,3,4,5,1,2,1,1,1])
val , count = np.unique(x , return_counts=True)
print(val)
print(count)
freq = np.argmax(count)
final = val[freq]
print(final)
```

```python
import numpy as np
mat = np.matrix('[4,1,9; 12,3,1; 4,5,6]')
stum = np.sum(mat)
rsum = np.sum(mat , axis=1)
csum = np.sum(mat , axis = 0)
print(stum)
print(rsum)
print(csum)
```

```python
import numpy as np

mat2 = np.array([[55,25] , [11 , 45]])
dsum = np.trace(mat2)
eigenvalues, eigenvectors = np.linalg.eig(mat2)
minv = np.linalg.inv(mat2)

mdet = np.linalg.det(mat2)
print(dsum)
print(eigenvalues)
print(eigenvectors)
print(minv)
print(mdet)
```

```python
import numpy as np
p1 = np.array([[1,2],[2,3]])
q1 = np.array([[4,5],[6,7]])
mul = np.dot(p1,q1)
cov = np.cov(p1.ravel(),q1.ravel())
print(mul)
print(cov)
```

```python
import numpy as np
import itertools
x = np.array ([[2,3,4] , [3,2,9]])
y = np.array([[1,5,0],[5,10,3]])
inner_prod = np.inner(x,y)
outer_prod = np.outer(x,y)
cartesian = list(itertools.product(x.ravel(),y.ravel()))
print(inner_prod)
print(outer_prod)
print(cartesian[:5])
print(cartesian[-5:])
```

```python
import numpy as np
array = np.array([[1,-2,3], [-4,5,-6]])
abs_val= np.abs(array)
print(abs_val)
parr = np.percentile(array , [25,50,75])
pcol = np.percentile(array , [25,50,75] , axis=0)
prow = np.percentile(array , [25,50,75] , axis=1)
print(parr)
print(pcol)
print(prow)
xf , mf , stf = np.mean(array) , np.median(array) , np.std(array)
print(xf , mf , stf)
```

```python
import numpy as np
array = np.array([10,52,62,16,16,54,453])
s_arr = np.sort(array)
ind_arr=np.argsort(array)

s_4_smallest = s_arr[:4]

l_5_largest = s_arr[-5:]
print("Sorted Array:", s_arr)
print("Indices of Sorted Array:", ind_arr)
print("4 Smallest Elements:", s_4_smallest)
print("5 Largest Elements:", l_5_largest)
```
